```python
#computes ammounts of gc in sequence
def oneComputeGCtotal(seq):
    gc_content = 0
    total_bases = len(seq)
    idx = 0
    while idx < len(seq):
        seq_idx = seq[idx]
        if seq_idx == 'C' or seq_idx == 'G':
            gc_content += 1
        idx += 1
    gc_cont_perc = float(gc_content) / float(total_bases)
    return gc_cont_perc
```

```python
def compute_AT_and_GC_percent(seq):
    gc_percent = oneComputeGCtotal(seq)
    at_percent = float(1) - gc_percent
    return [gc_percent,at_percent]
```

```python
#For an entire base probability matrix, can't specify start or end
def find_single_strandedness(base_pair_probability_matrix): #This was taken from Andrew's code, for some reason it's not working as intended... possible that diagonals means something different in Vienna
    assert base_pair_probability_matrix.ndim == 2, "matrix must be two dimensional"
    assert base_pair_probability_matrix.shape[0] == base_pair_probability_matrix.shape[1], "matrix must be square"

    # sum_of_diagonal = np.sum(np.diag(base_pair_probability_matrix))
    # length_of_sequence = base_pair_probability_matrix.shape[0]
    # return sum_of_diagonal/length_of_sequence
    top_triangle_probability_matrix = np.triu(base_pair_probability_matrix, 1)#zeroes out the main diagonal of the matrix plus all the diagonals below it
    probability_for_each_nucleotide = np.sum(top_triangle_probability_matrix, axis=0)#Give us the probability of each nucleotide that it's bonding to smth
    probability_sum = np.sum(probability_for_each_nucleotide) #sums up all the probabilities, to give us the total percent
    #print(probability_sum)
    return 1-(probability_sum/(base_pair_probability_matrix.shape[0]-1)) #We then divide by the size of the array-1 (because the bpp of viennaRNA adds an extra row/col for nice formatting) also subtract by one to get the prob of being unpaired
```

```python
def calculate_kmers(sequence, start, end, kmer_lengths=[3, 4, 5]):
    
    subsequence = sequence[start:end]

    kmer_counts = {}

    for k in kmer_lengths:
        for i in range(len(subsequence) - k + 1):
            kmer = subsequence[i:i+k]
            if kmer in kmer_counts:
                kmer_counts[kmer] += 1
            else:
                kmer_counts[kmer] = 1

    return kmer_counts
```

```python
#Can specify start or end with this version
#The index starts at 0, code auto modifies your input to work with the bpp that starts at index 1...
def probability_of_being_unpaired(base_pair_probability_matrix, start, end): #Gets single strandedness for nucleotide sequence from start to end index (end nucleotide not included)
    assert base_pair_probability_matrix.ndim == 2, "matrix must be two dimensional"
    assert base_pair_probability_matrix.shape[0] == base_pair_probability_matrix.shape[1], "matrix must be square"

    top_triangle_probability_matrix = np.triu(base_pair_probability_matrix, 1)#zeroes out the main diagonal of the matrix plus all the diagonals below it
    #Slice the matrix to do easier sum stuff (might be pointless but hey why not)
    sub_matrix = top_triangle_probability_matrix[start+1:end+1, :]#Starts at index 1, so we have to increment everything by 1 (0th nucleotide is 1st row in the matrix and so on)
    probability_for_each_nucleotide = np.sum(sub_matrix, axis=0)#Give us the probability of each nucleotide that it's bonding to smth
    probability_sum = np.sum(probability_for_each_nucleotide) #sums up all the probabilities, to give us the total percent
    #print(probability_sum) 
    return 1-(probability_sum/(end-start))
```

```python
def test_for_enclosed_structure(structure, start_ind, length):
    end_ind = start_ind + length
    if(start_ind==0):
        raise Exception("Start index cannot be 0")
    start_ind -=1 #This is because we are going to start checking the structure before the start index for an enclosed structure

    increment_count = 0
    limit = 10

    # print(structure[start_ind:end_ind+1])

    while start_ind !=0 and structure[start_ind] != '(' and increment_count < limit:
        start_ind -=1
        increment_count +=1
        # print('decrementing left')
        # print(structure[start_ind:end_ind+1])

    if(structure[start_ind]!='('):
        # print("Not Valid")
        return False
    
    increment_count = 0

    while end_ind != len(structure)-1 and structure[end_ind] != ')' and increment_count<limit:
        end_ind +=1
        increment_count +=1
        # print("increasing right")
        # print(structure[start_ind:end_ind+1])
        
    if structure[end_ind] == ')':
        # print("Valid")
        return True
    # else:
    #     print("Not Valid")
    #     return False
```

```python
def test_validity_of_rbs(riboswitch_sequence, riboswitch_structure):
    valid = False 
    index = None
    aug_limit = 15 #Relatively close to 10 + len of rbs. Can change later depending on needs

    rbs_index = riboswitch_sequence.find("AGGAGA") #This is the rbs sequence (ribosomal binding site, where the ribosome will bind to start transcribing the reporter gene)
    #print(rbs_index)

    if(rbs_index > 40 and rbs_index<110): #Shouldn't be located near the beginning or end unless smth wrong has occured
        if(test_for_enclosed_structure(riboswitch_structure, rbs_index, 6)):
            if(riboswitch_sequence.find("AUG", rbs_index)-rbs_index<=aug_limit): #There should be a start codon, or AUG, after the rbs. Check to see if there is one close by
                valid = True
                index = rbs_index
    return [valid,index]
    
```

```python
def test_valid_notation(d): #just sees that every opening parenthesis has a closing parethesis (i.e is valid notation)
    q = deque()
    structure_present = False
    for let in d:
        if let == '(':
            q.append(let)
            structure_present = True 
        elif let == ')':
            if (len(q)==0):
                return 1 #if extra closed parethesis
            q.pop()
    if(len(q)==0):
        if structure_present is True:
            return 0 #if valid
        else:
            return 1 #There is no structure at all present. That isn't a loop so let's add an open parenthesis to see if the loop got shifted over
    return 2 #if extra open parethesis
```

```python
#NOTE: actually figure out which nueclotides are the stem loop region
def stem_loop_eval(s): #Stem loop region is 35-48 nucleotide inclusive
    start = 65
    end = 79
    strand_sub = s[start:end]
    ind = 1
    limit = 3 #Limit is 3 because seems like a good num i suppose
    result = test_valid_notation(strand_sub)
    #print(strand_sub)
    while result!=0 and ind<=limit: #there might be off by 1 errors, so we are gonna add up to the limit that many nueclotides to the sequence to try to get a valid sequence
        if result==1:
            start-=1
        else:
            end+=1
        strand_sub = s[start:end]
        #print(strand_sub) 
        result = test_valid_notation(strand_sub)
        #print(ind)
        ind+=1
    if result == 0:
        return True, start, end
    return False, -1, -1
        
```

```python
def trigger_target_binding(struct, len): #len is length of target or trigger (they should be the same size)
    num_binds = 0
    open_trigger_nucleotides = 0
    # print(struct[0:60])
    for i in range(len): #Trigger should be the len of target and 0+len nucleotides into the structure
        if struct[i+len] == '(': 
            open_trigger_nucleotides+=1
        elif struct[i+len] == ')': 
            if open_trigger_nucleotides>0:
                open_trigger_nucleotides-=1
            else:
                num_binds +=1
    percent_bind = num_binds/len
    # print(struct[0:len])
    # print(struct[len:len*2])
    # print(num_binds)
    return percent_bind
```

```python
def calculate_negative_categories(riboswitch_data, name_of_riboswitch_dataframe, switch_col, trigger_length, target_data, name_of_target_dataframe, target_col, csv_names, reporter_seq, 
                    reporter_table, label, location_col = -1, subname_col = -1):
    
    print("Started method")
    riboswitch_data = riboswitch_data.dropna()
    target_data = target_data.dropna()
    trigger_data = trigger_generator(target_data, target_col, location_col)
    trigger_sequences_series = trigger_data['trigger_sequence']
    display(target_data)
    display(trigger_data)

    ribo_size = len(riboswitch_data)
    switch_size = len(target_data)

    vArr = [[0 for i in range (len(col_names))] for j in range (ribo_size*switch_size*switch_size)] #fix later lol
    print(len(vArr))
    #test = deepcopy(vArr)
    full_table = [[] for i in range(1)]
    #0 is salt
    #1 is temperature
    #2 is switch row
    #3 is target row
    #4 is target location
    #5 is toehold switch sequence
    #6 is target sequence
    #7 is switch mfe struct
    #8 is target mfe struct
    #9 is switch energy
    #10 is target energy
    #11 is valid stem loop
    #12 is stem loop sequence
    #13 is stem loop structure
    #14 is stem loop energy
    #15 is complex nucelotides
    #16 is complex structure
    #17 is complex visualization
    #18 is complex energy
    #19 is complex gc count
    #20 is switch gc count
    #21 is target gc count
    #22 is switch ensemble defect
    #23 is target ensemble defect
    #24 is complex ensemble defect
    #25 is switch single strandness
    #26 is target single strandness
    #27 is complex single strandness
    for i in range(len(salt_concentrations_list)):
        print('new sheet')
        md.salt = salt_molarity = salt_concentrations_list[i]

        print("new concetration now", salt_molarity, flush=True)


        ribocopy_df = pd.DataFrame()
        targetcopy_df = pd.DataFrame()

        ribocopy_df[riboswitch_table_names] = make_riboswitch_table(riboswitch_data, switch_col, trigger_length, trigger_sequences_series,subname_col)
        targetcopy_df[target_table_names] = make_negative_target_table(target_data, target_col, location_col)
       
        print("preliminary data done and dusted", flush=True)

        for j in range(len(ribocopy_df)):
            x = j
            t = j%switch_size
            
            print("New row of table:", x)
            print("Trigger is:", t)
            if(j%switch_size==0):
                print("New riboswitch:", x/switch_size)
            switch_struct = ribocopy_df.at[x,'switch_structure']
            switch_seq = ribocopy_df.at[x,'riboswitch_sequence']
            switch_subname = ribocopy_df.at[x,'riboswitch_subcategory']
            riboswitch_row = j//switch_size
            trigger_seq = trigger_data.at[t,'trigger_sequence']

            for y in range(switch_size):
                arr_loc = x * switch_size + y
                print("Arr_loc:", arr_loc)
                print("new target:", y)
                data_list = []
                if (y==t):
                    print("T and Y collides",y, t)
                    data_list = [None] * len(col_names)
                    fill_matrix(arr_loc,vArr, data_list)
                    continue
                target_seq = targetcopy_df.at[y,'target_sequence']
                target_struct = targetcopy_df.at[y,'target_structure']

                #data 0-10
                #adding salt, temperature, switch row, target row, target location, toehold switch sequence, target sequence, switch mfe structure, target mfe structure, switch energy, and target energy to list
                data_list.extend([label,salt_molarity, temp, name_of_riboswitch_dataframe, name_of_target_dataframe, switch_subname,riboswitch_row, t, y, 
                                  targetcopy_df.at[y, 'target_location'],trigger_data.at[t, 'Location'], switch_seq, 
                                  trigger_seq, target_seq, target_seq+trigger_seq, reporter_seq, switch_struct, target_struct, ribocopy_df.at[x,'switch_energy'], 
                                  targetcopy_df.at[y,'target_energy'],ribocopy_df.at[x,'trigger_single_strandedness'], ribocopy_df.at[x,'trigger_structure'],
                                  ribocopy_df.at[x, 'switch_kmers'], targetcopy_df.at[y, 'target_kmers'], ribocopy_df.at[x,'rbs_validity'], ribocopy_df.at[x,'rbs_location']])


                #This is scuffed, there should be a better way but this will do for now
                #This evaluates the stem loop and its free energy, if it's valid
                outcome = stem_loop_eval(switch_struct)
                #Add if the loop is valid (#11)
                validity = outcome[0]
                data_list.append(validity)

                #now add the loop sequence, structure, and energy if applicable (#12-14)

                if(validity==False): #if not valid from stem loop test 
                    data_list.extend(['N/A', 'N/A', 'N/A'])
                else:
                    #we use substrings of the switches sequence and structure to get the loop structure and sequence
                    sub_seq = switch_seq[outcome[1]:outcome[2]]
                    sub_struct = switch_struct[outcome[1]:outcome[2]]
                    #print(sub_struct)
                    data_list.extend([sub_seq, sub_struct, RNA.fold_compound(sub_seq,md).mfe()[1]])
                
                
                #Add the complex sequence + subsequence of reporter gene (#15)
                short_reporter = reporter_seq[:10] #we can't use the entire subseqence because that blows up everything with pf. Only 10 base pairs
                data_list.append(target_seq+switch_seq+short_reporter)

                #Add the reporter gene to the switch
                full_riboswitch = switch_seq+short_reporter
                #Current NOTE: We don't need to replace the T's with U anymore so lol
                # Old NOTE SOME OF THE SEQUENCES HAVE T's IN THEM INSTEAD OF U's... WE GOTTA REPLACE THEM, Original is still listed in table
                # replaced_target_seq = target_seq.replace('T','U')
                # new_switch_seq = full_riboswitch.replace('T','U')
                # replaced_short_reporter = short_reporter.replace('T','U')
    
                # print(len(target_seq+new_switch_seq))

                fc = RNA.fold_compound(target_seq+'&'+switch_seq+'&'+short_reporter, md)
                complex_sequence = target_seq+switch_seq+short_reporter

                (v_complex_struct, v_complex_mfe) = fc.mfe()

                #Add the structure, visualization structure, mfe energy, complex gc count, switch gc count, and target gc count (#16-21)
                #NOTE: See if we want to compute AT GC for the WHOLE complex or with just the short reporter later 
                [complex_gc, complex_at] = compute_AT_and_GC_percent(complex_sequence)
                data_list.extend([v_complex_struct, v_complex_struct.replace('+','&'), v_complex_mfe, complex_gc, complex_at,
                                  ribocopy_df.at[x,'switch_gc'],ribocopy_df.at[x, 'switch_at'], targetcopy_df.at[y, 'target_gc'], targetcopy_df.at[y, 'target_at']])
                
                #ensemble defect/parititon function calculations
                #We add the switch, target, and complex ensemble defect here (#22-24)
                (complex_propensity, complex_ensemble_energy) = fc.pf()
                data_list.extend([ribocopy_df.at[x, 'switch_ensemble_defect'], targetcopy_df.at[y, 'target_ensemble_defect'], fc.ensemble_defect(v_complex_struct)])

                #Calculate and add the switch, target, and complex single strandedness (#24-27)
                #NOTE: Removing the below for now because pf doesn't work well with big sequences, it's broken anyways
                complex_bpp = fc.bpp()
                np_complex = np.asarray(complex_bpp)
                data_list.extend([ribocopy_df.at[x, 'switch_single_strandedness'],targetcopy_df.at[y, 'target_single_strandedness'], find_single_strandedness(np_complex)])

                complex_kmer_dict = calculate_kmers(complex_sequence, 0, len(complex_sequence))
                data_list.extend([complex_kmer_dict])

                data_list.extend(reporter_table[i])
                
                length_of_trigger = 30
                data_list.extend([trigger_target_binding(v_complex_struct,length_of_trigger)])

                fill_matrix(arr_loc, vArr, data_list)
                #fill_matrix(arr_loc, test, data_list)

                #UNCOMMENT BELOW
                full_table.append(data_list)
                #UNCOMMENT ABOVE
        concentration_data = pd.DataFrame(vArr)
        concentration_data.columns = col_names
        concentration_data = concentration_data[finished_data_names]
        display(concentration_data)
        concentration_data.to_csv(csv_names[i])
        del concentration_data
        del ribocopy_df
        del targetcopy_df
        # print("this is the array")
        # display(pd.DataFrame(vArr))
        # print('this is the full_table')
        # display(pd.DataFrame(full_table))

    

    # full_table = full_table[1:] #To get rid of the first empty row
    # #full_table.insert(0, col_names)
    # #print(vArr)
    # results = pd.DataFrame(full_table)
    # results.columns = col_names
    # results = results[finished_data_names]
    results = 1
    #testing = pd.DataFrame(test)
    return results
```
